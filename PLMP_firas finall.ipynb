```python
def clean_data(df):
    df = df.drop(index=[0, 1])
    df.columns = [df.iloc[0][i] + '_' + str(i) for i in range(len(df.columns))]
    df = df[1:]
    df.reset_index(drop=True, inplace=True)
    return df
```

```python
def prefix_columns(df, prefix):
    new_df = df.copy()
    new_df.columns = [new_df.columns[0]] + [f"{prefix}_{col}" for col in new_df.columns[1:]]
    
    return new_df
```

```python
Autobahnen = clean_data(Autobahnen)
Wasserwege = clean_data(Wasserwege)
Bundesstraßen = clean_data(Bundesstraßen)
Fläche = clean_data(Fläche)
Gewerbegebiete = clean_data(Gewerbegebiete)
Krankenhäuser = clean_data(Krankenhäuser)
Privathaushalte = clean_data(Privathaushalte)
Schienenwege = clean_data(Schienenwege)
Schulen = clean_data(Schulen)
Unternehmen = clean_data(Unternehmen)
# data10 = clean_data(data10)
```

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
```

```python
Flächen_numeric = Fläche[['Fläche_2G_6', 'Fläche_4G_7', 'Fläche_5G DSS_8', 'Fläche_5G_9']]
Flächen_numeric = Flächen_numeric.replace({',': '.'}, regex=True).astype(float)

scaler = StandardScaler()
Flächen_scaled = scaler.fit_transform(Flächen_numeric)
```

```python

inertia = []
for k in range(1, 9):  
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(Flächen_scaled)
    inertia.append(kmeans.inertia_)

# Plotting the inertia to identify the "elbow"
plt.plot(range(1, 9), inertia)
plt.xlabel('Number of Clusters')
plt.ylabel('Inertia')
plt.title('Elbow Method')
plt.show()
```

```python
kmeans = KMeans(n_clusters=3,  init='k-means++', random_state=42)
Fläche['Cluster'] = kmeans.fit_predict(Flächen_scaled)

# Show the dataframe with cluster assignments
print(Fläche[['AGS_0', 'Fläche_Name_1', 'Cluster']])
```

```python
X = data1[['Schulen_mbit1000', 'Unternehmen_mbit1000' , 'Gewerbegebiete_mbit1000' , 'Privathaushalte_≥ 1000 Mbit/s_12']]
y = data1['wieOft_Landkreis']
```

```python
for col in X.columns:
    X[col] = X[col].astype(str).str.replace(',', '.').astype(float)
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split

# erst in Training und Test aufteilen, dann nur auf den Trainingsdaten trainieren
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)

regr = RandomForestRegressor(max_depth=5, random_state=0)
regr.fit(X_train, y_train)
```

```python
y_pred = regr.predict(X_test)
```

```python
y_pred
```

```python
# Aktuelle_daten.csv fehlt im Repo. Die Werte darin stammen aus dem neueren Breitbandatlas-Stand
# im Ordner "data2für trainieren" (gleiche Zahlen wie in der alten Ausgabe), deshalb hier neu aufgebaut.
spalten = {
    'Schulen': 'Schulen_mbit1000',
    'Unternehmen': 'Unternehmen_mbit1000',
    'Gewerbegebiete': 'Gewerbegebiete_mbit1000',
    'Privathaushalte': 'Privathaushalte_≥ 1000 Mbit/s_12',
}
aktuelle_daten = None
for name, neuer_name in spalten.items():
    df = prefix_columns(clean_data(pd.read_csv(f'data2für trainieren/{name}.csv')), name)
    df = df[df[f'{name}_Verwaltungsebene_2'] == '3 - Kreis']
    df = df[['AGS_0', f'{name}_≥ 1000 Mbit/s_12']].rename(columns={f'{name}_≥ 1000 Mbit/s_12': neuer_name})
    df[neuer_name] = df[neuer_name].astype(str).str.replace(',', '.').astype(float)
    if aktuelle_daten is None:
        aktuelle_daten = df
    else:
        aktuelle_daten = pd.merge(aktuelle_daten, df, on='AGS_0', how='inner', validate='one_to_one')
```

```python
aktuelle_daten 
```

```python
X_to_predict = aktuelle_daten[['Schulen_mbit1000', 'Unternehmen_mbit1000', 'Gewerbegebiete_mbit1000', 'Privathaushalte_≥ 1000 Mbit/s_12']]
```

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

# Vergleich: immer den Mittelwert aus den Trainingsdaten vorhersagen
y_baseline = np.full(len(y_test), y_train.mean())

ergebnis = pd.DataFrame({
    'MAE': [mean_absolute_error(y_test, y_baseline), mean_absolute_error(y_test, y_pred)],
    'RMSE': [np.sqrt(mean_squared_error(y_test, y_baseline)), np.sqrt(mean_squared_error(y_test, y_pred))],
    'R²': [r2_score(y_test, y_baseline), r2_score(y_test, y_pred)],
}, index=['Mittelwert aus Training', 'Random Forest'])
ergebnis
```
